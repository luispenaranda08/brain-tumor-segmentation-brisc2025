# **2. Revisión de Literatura**

## 2.1 Evolución de los Métodos de Segmentación en Imagen Médica

La segmentación de imágenes médicas ha evolucionado significativamente durante la última
década, pasando de enfoques basados en características artesanales (*handcrafted features*)
hacia arquitecturas de aprendizaje profundo capaces de extraer representaciones jerárquicas
directamente desde los datos. Esta transición fue impulsada en gran medida por la introducción
de las Redes Neuronales Convolucionales (CNN) como estándar de facto en visión por
computadora, y posteriormente por la emergencia de los Vision Transformers como alternativa
capaz de capturar dependencias globales de largo alcance {cite}`nawaz2025review`.

En el contexto específico de tumores cerebrales, la literatura reporta un avance sostenido
desde métodos clásicos de segmentación basados en umbrales y crecimiento de regiones, pasando
por enfoques de Machine Learning con SVM y Random Forest, hasta los modelos de Deep Learning
actuales que dominan los benchmarks en los principales desafíos del área como BraTS
{cite}`menze2014multimodal`.

---

## 2.2 U-Net y sus Variantes: El Paradigma Encoder-Decoder

La arquitectura U-Net, propuesta por Ronneberger et al. {cite}`ronneberger2015unet`, estableció
en 2015 el paradigma que define la segmentación médica moderna. Su diseño encoder-decoder
simétrico con conexiones de salto (*skip connections*) resuelve el problema fundamental de
la segmentación semántica densa: preservar la información espacial de alta resolución mientras
se extrae contexto semántico en profundidad. La simplicidad de su diseño, combinada con su
capacidad de entrenarse con conjuntos de datos reducidos gracias al uso extensivo de
augmentación de datos, la convirtió rápidamente en la arquitectura de referencia en imagen
médica.

Una de las extensiones más influyentes de U-Net fue Attention U-Net, propuesta por Oktay et
al. {cite}`oktay2018attention`. Esta variante introduce puertas de atención (*attention gates*)
en las conexiones de salto, lo que permite al modelo aprender implícitamente a suprimir
activaciones irrelevantes y enfocarse en las regiones de interés sin requerir módulos externos
de localización. En la tarea de segmentación de páncreas en CT — uno de los problemas más
difíciles en imagen médica por la variabilidad morfológica del órgano — Attention U-Net
demostró mejoras consistentes sobre U-Net estándar manteniendo eficiencia computacional. Este
mecanismo de atención en skip connections es directamente relevante para nuestro problema,
dado que los tumores cerebrales presentan morfologías igualmente variables e irregulares.

---

## 2.3 Mecanismos de Atención en Redes Convolucionales

Más allá de las puertas de atención en skip connections, la literatura ha explorado módulos
de atención aplicados directamente sobre los mapas de características del encoder. El módulo
CBAM (*Convolutional Block Attention Module*) combina atención de canal y atención espacial
de forma secuencial, permitiendo al modelo recalibrar tanto qué características son relevantes
como dónde se ubican en la imagen {cite}`woo2018cbam`. Este tipo de atención ligera ha
demostrado mejoras consistentes en tareas de clasificación y segmentación con un costo
computacional marginal.

En el contexto de tumores cerebrales, Prayogo et al. {cite}`prayogo2026hayabusanet` propusieron
HayabusaNet, una arquitectura de clasificación que integra un módulo de atención híbrida
combinando CBAM para prominencia local y mecanismos de self-attention para contexto global.
Aunque su tarea es clasificación y no segmentación, los resultados demuestran que la
combinación de atención local y global es beneficiosa para discriminar entre tipos tumorales
con morfologías variables, lo que fundamenta su aplicación dentro de arquitecturas encoder-decoder
para segmentación.

## 2.4 Vision Transformers en Segmentación Médica

La introducción de los Transformers en visión por computadora abrió una nueva era en
segmentación médica. A diferencia de las CNN, cuyas operaciones de convolución tienen una
receptividad local inherente, los Transformers modelan dependencias entre todos los píxeles
o parches de la imagen simultáneamente a través de mecanismos de autoatención, lo que
les permite capturar contexto global desde las primeras capas del modelo.

TransUNet, propuesto por Chen et al. {cite}`chen2021transunet`, fue uno de los primeros trabajos
en integrar Vision Transformers en una arquitectura de segmentación médica. Su diseño híbrido
utiliza una CNN para extraer mapas de características de alta resolución, que luego son
tokenizados y procesados por un encoder basado en ViT para capturar contexto global. El decoder,
inspirado en U-Net, combina las características codificadas por el Transformer con las
características CNN de alta resolución mediante skip connections, logrando localización precisa.
TransUNet ha demostrado rendimiento superior en tareas de segmentación multi-órgano y
cardíaca, y su versión 3D publicada en Medical Image Analysis en 2024 supera a nnUNet en la
mayoría de benchmarks evaluados {cite}`chen2024transunet`.

Swin-UNet {cite}`cao2022swinunet` llevó este paradigma un paso más allá al proponer la
primera arquitectura de segmentación médica basada puramente en Transformers, sin ningún
componente convolucional. Utilizando el Swin Transformer jerárquico con ventanas desplazadas
como encoder, y un decoder simétrico también basado en Swin Transformer con una capa de
expansión de parches para el upsampleo, Swin-UNet demuestra que una red U-shaped basada
puramente en Transformers puede superar a métodos con convolución completa o combinación de
Transformer y convolución en tareas de segmentación multi-órgano y cardíaca.

## 2.5 Estado del Arte en Segmentación de Tumores Cerebrales

La segmentación de tumores cerebrales ha sido impulsada históricamente por el desafío BraTS
(*Brain Tumor Segmentation*), organizado anualmente desde 2012 por la sociedad MICCAI
{cite}`menze2014multimodal`. BraTS se enfoca principalmente en gliomas y utiliza imágenes
multimodales (T1, T1-CE, T2, FLAIR) en formato 3D. Los mejores modelos en BraTS han
alcanzado Dice scores superiores a 0.90 para tumor completo, con arquitecturas híbridas
CNN-Transformer dominando los rankings más recientes {cite}`nawaz2025review`.

Sin embargo, la literatura presenta limitaciones importantes que motivan el presente trabajo.
Primero, la mayoría de estudios se enfocan en gliomas de alto grado, dejando subrepresentados
los meningiomas y tumores hipofisarios {cite}`nawaz2025review`. Segundo, los datasets más
usados — BraTS, Figshare, datasets de Kaggle — presentan problemas de desbalance de clases,
inconsistencias en anotaciones o falta de diversidad en condiciones de adquisición
{cite}`fateh2026brisc`. Tercero, y más relevante para este trabajo, prácticamente ningún estudio
analiza el rendimiento de segmentación de forma desagregada por plano anatómico de adquisición
(axial, coronal, sagital) en imágenes 2D de MRI T1, a pesar de que la morfología tumoral
percibida varía considerablemente entre planos.

Una revisión reciente publicada en npj Precision Oncology ofrece un panorama comprehensivo
del estado del arte en análisis de tumores cerebrales con Deep Learning, concluyendo que los
modelos más robustos son aquellos que integran mecanismos de atención multi-escala capaces
de adaptarse a la alta variabilidad morfológica inter-paciente e inter-tipo tumoral
{cite}`elhalawani2025review`. Esta conclusión es consistente con la hipótesis de nuestro
trabajo.

## 2.6 Trabajos Recientes sobre el Dataset BRISC 2025

Desde su publicación, el dataset BRISC ha atraído varios trabajos que lo utilizan principalmente
para tareas de clasificación. Bindemann y Schanze {cite}`bindemann2026search` realizaron una
exploración sistemática de 682 configuraciones de CNN sobre BRISC, demostrando que modelos
con la mitad de parámetros pueden alcanzar precisiones comparables a arquitecturas más pesadas,
con un máximo de 0.985 en clasificación. En una línea similar, Prabhakar y Swamy
{cite}`prabhakar2026deep` evaluaron CNNs personalizadas con pipelines robustos de preprocesamiento
sobre BRISC, identificando que arquitecturas ligeras como EfficientNet-b0 alcanzan precisiones
superiores al 98% en clasificación multiclase.

Es importante destacar que estos trabajos abordan clasificación a nivel de imagen, una tarea
fundamentalmente distinta a la segmentación pixel a pixel que es el objeto del presente
proyecto. La ausencia de estudios de segmentación con análisis por plano anatómico sobre BRISC
confirma el gap que este trabajo se propone cubrir.

## 2.7 Desplazamiento de Dominio en Imagen Médica

Un desafío relevante identificado durante el análisis exploratorio de datos de este proyecto
es la presencia de un desplazamiento de distribución estadísticamente significativo entre los
conjuntos de entrenamiento y prueba de BRISC. Esta observación está respaldada por la
literatura más reciente sobre imagen médica con Deep Learning.

Kakon et al. {cite}`kakon2026improving` demostraron que los modelos de análisis de MRI sufren
degradaciones en su rendimiento cuando se enfrentan a variaciones en los protocolos de
adquisición o en las características de los equipos, proponiendo un marco de regularización
de estabilidad en el espacio de características (*Feature Space Stability Regularization*)
como estrategia de mitigación. En un marco más amplio, Musa et al. {cite}`musa2026systematic`
realizaron una revisión sistemática de los desplazamientos entre poblaciones en imagen médica
con Deep Learning, identificando que la variación en condiciones de adquisición y
características demográficas representa el principal obstáculo para la generalización de los
modelos. Ambos trabajos sustentan la importancia de reportar y discutir el comportamiento del
modelo bajo las condiciones de evaluación real del dataset, incluyendo las diferencias de
distribución entre splits, como se analiza en este proyecto.

## 2.8 Limitaciones de los Datasets Existentes y la Propuesta BRISC

La calidad y características de los datasets de entrenamiento son tan determinantes para el
rendimiento de los modelos como las propias arquitecturas. Los principales datasets públicos
para segmentación de tumores cerebrales presentan limitaciones documentadas que BRISC fue
diseñado para superar {cite}`fateh2026brisc`.

BraTS depende de datos preprocesados y estandarizados que pueden no reflejar la variabilidad
real en protocolos de adquisición clínica, y se enfoca exclusivamente en gliomas, limitando
la generalización a otros tipos tumorales {cite}`menze2014multimodal`. El dataset Cheng
(*Figshare*) sufre de desbalance de clases y diversidad limitada en condiciones de imagen.
Muchos datasets de Kaggle ampliamente utilizados presentan inconsistencias en las anotaciones
que afectan negativamente el entrenamiento {cite}`fateh2026brisc`.

BRISC aborda estas limitaciones ofreciendo 6.000 imágenes MRI T1 con contraste, con
distribución balanceada entre cuatro clases — glioma, meningioma, pituitary y no-tumoral —
y tres planos anatómicos, con máscaras de segmentación pixel a pixel validadas por radiólogos
y médicos certificados. El coeficiente Dice de 0.924 entre anotaciones iniciales y verificadas
por expertos establece un estándar de calidad superior al de la mayoría de datasets públicos
disponibles {cite}`fateh2026brisc`. Su publicación en Scientific Data (Nature Portfolio) en
2026 lo posiciona como el dataset de referencia más reciente y riguroso para esta tarea.

## 2.9 Síntesis y Posicionamiento del Presente Trabajo

La revisión de la literatura permite identificar con claridad el espacio de contribución de
este trabajo. Desde el punto de vista arquitectónico, el paradigma híbrido CNN-Transformer
con mecanismos de atención multi-escala representa el estado del arte más prometedor para
segmentación de tumores cerebrales en imágenes 2D de MRI T1 {cite}`chen2021transunet`,
{cite}`cao2022swinunet`. Los módulos de atención de canal y espacial, como CBAM, han
demostrado su utilidad en el contexto de tumores cerebrales con morfología variable
{cite}`woo2018cbam`, {cite}`prayogo2026hayabusanet`. Desde el punto de vista de datos, BRISC
2025 ofrece el conjunto de referencia más completo y riguroso disponible para esta tarea
{cite}`fateh2026brisc`. Y desde el punto de vista analítico, el análisis de rendimiento
desagregado por plano anatómico representa un gap real en la literatura que este trabajo se
propone cubrir, documentado tanto por la ausencia de dicho análisis en los trabajos existentes
sobre BRISC como por la evidencia de desplazamiento de dominio entre splits {cite}`kakon2026improving`,
{cite}`musa2026systematic`.
